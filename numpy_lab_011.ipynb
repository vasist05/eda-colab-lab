```python
#1.     Create a 3×4 NumPy array with integers from 0 to 11. Transpose the array using .T and verify the shape becomes 4×3. Print both arrays and observe the structural changes. (Transposing 2D Matrix)
import numpy as np
arr=np.arange(0,12).reshape(3,4)
transarr=arr.T
print(transarr)
print()
print(arr)
print("shape of arr is:",arr.shape)
print()
print("shape of arr is:",transarr.shape)
```

```python
#2. Create a 3D array of shape (2, 3, 4) with sequential numbers using np.arange() and reshape. Transpose it using .transpose(1, 0, 2) to reorder the shape as (3, 2, 4). (Transpose a 3-D Array)
# Print the new shape and verify that data was rearranged accordingly.
arr=np.arange(0,24).reshape(2,3,4)
print(arr)
print()
transarr=arr.transpose(1,0,2)
print(transarr)
print()
print("shape of transarr is:",transarr.shape)
print()

```

```python
#4. 	Generate a 6×2 array. Transpose it using .T, and then apply np.swapaxes(array, 0, 1) on the original array.
#Show that both results are the same and explain why. (Compare Transpose and Swapaxes)
arr=np.arange(0,12).reshape(6,2)
arr1=arr.T
arr2=np.swapaxes(arr,0,1)
print(arr1)
print()
print(arr2)
print(arr1==arr2)
```

```python
#5. 	Create a 3D array with shape (2, 4, 3). Use np.swapaxes to swap axes 0 and 2, and print the new shape.
# Check how the data has been reorganized by printing slices from both arrays. (Swapaxes on a 3D Array)
arr=np.arange(1,25).reshape(2,4,3)
print(arr)
arr2=np.swapaxes(arr,2,0)
print(arr2)
print("shape of arr2 is:",arr2.shape)
print()
print(arr[1,1])
print(arr2[1,1])
```

```python
#Generate a 6×4 NumPy array of random integers between 50 and 100, then find row-wise means, column-wise means, column-wise standard deviations,
# and determine the column with the highest standard deviation.
arr=random.randint(low=50,high=101,size=(6,4))
print(arr)
rmean=arr.mean(axis=1)
print(rmean)
cmean=arr.mean(axis=0)
print(cmean)
rstd_dev=arr.std(axis=1)
print(rstd_dev)
cstd_dev=arr.std(axis=0)
print(cstd_dev)
print("the column with higest std deviation is:",cstd_dev.argmax())
#argmax() gives us index with max element
```

```python
#Create a 6×6 NumPy array of random floats from a normal distribution (mean 0, standard deviation 1),
# then for each column calculate (max - min) and identify the column with the largest range.
import numpy as np
rng=np.random.default_rng()
arr=rng.standard_normal(size=(6,6))
cmax=arr.max(axis=0)
cmin=arr.min(axis=1)
#print("max value in each col is:",cmax)
#print("min value in each col is:",cmin)
range=cmax-cmin
print("the range of cols is :",range)
print("the column with largest range is:",range.argmax())
```
