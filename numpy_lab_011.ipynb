```python
#Create a 6×6 NumPy array of random floats from a normal distribution (mean 0, standard deviation 1),
# then for each column calculate (max - min) and identify the column with the largest range.
import numpy as np
rng=np.random.default_rng()
arr=rng.standard_normal(size=(6,6))
cmax=arr.max(axis=0)
cmin=arr.min(axis=1)
#print("max value in each col is:",cmax)
#print("min value in each col is:",cmin)
range=cmax-cmin
print("the range of cols is :",range)
print("the column with largest range is:",range.argmax())
```

```python

import pandas as pd

# 1. Create the DataFrame
data = {
    "Emp_ID": ["E101", "E102", "E103", "E104", "E105", "E106", "E107", "E108"],
    "Department": ["IT", "HR", "Sales", "IT", "Finance", "Sales", "IT", "HR"],
    "Age": [25, 32, 29, 41, 36, 27, 45, 39],
    "Attendance": [92, 78, 88, 95, 72, 81, 68, 90],
    "Salary": [42000, 55000, 48000, 72000, 65000, 45000, 85000, 62000]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# 2. Attendance greater than 85%
print("\nAttendance greater than 85%:")
print(df[df["Attendance"] > 85])

# 3. Employees in IT department
print("\nEmployees in IT department:")
print(df[df["Department"] == "IT"])

# 4. Attendance > 80 AND Salary > 50000
print("\nAttendance > 80% AND Salary > ₹50,000:")
print(df[(df["Attendance"] > 80) & (df["Salary"] > 50000)])

# 5. Employees in IT or Finance using isin()
print("\nEmployees in IT or Finance:")
print(df[df["Department"].isin(["IT", "Finance"])])

# 6. Attendance not less than 75%
print("\nAttendance not less than 75%:")
print(df[df["Attendance"] >= 75])

```

```python

import pandas as pd

# 1. Create the DataFrame
data = {
    "Customer_ID": ["C101", "C102", "C103", "C104", "C105",
                    "C106", "C107", "C108", "C109", "C110"],
    "Age": [22, 35, 28, 42, 31, 24, 48, 39, 27, 52],
    "Annual_Spending": [12000, 28000, 45000, 72000, 55000,
                        18000, 95000, 68000, 32000, 110000]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# 2 & 3. Create spending categories using pd.cut()
bins = [0, 20000, 40000, 70000, float("inf")]
labels = ["Low", "Medium", "High", "Very High"]

df["Spending_Level"] = pd.cut(
    df["Annual_Spending"],
    bins=bins,
    labels=labels,
    right=False
)

print("\nDataFrame after binning:")
print(df)

# 4. Customers belonging to High spending category
print("\nHigh spending customers:")
print(df[df["Spending_Level"] == "High"])

# 5. Number of customers in each category
print("\nNumber of customers in each spending category:")
print(df["Spending_Level"].value_counts())

```

```python

import pandas as pd

# 1. Create the DataFrame
data = {
    "Movie_ID": ["M101", "M102", "M103", "M104", "M105",
                 "M106", "M107", "M108", "M109"],
    "Genre": ["Action", "Drama", "Comedy", "Action", "Drama",
              "Comedy", "Action", "Drama", "Comedy"],
    "Language": ["English", "Hindi", "Telugu", "Hindi", "English",
                 "Hindi", "Telugu", "Telugu", "English"],
    "Rating": [4, 5, 3, 4, 5, 2, 5, 4, 3]
}

df = pd.DataFrame(data)

# 2. Convert Genre and Language to categorical
df["Genre"] = df["Genre"].astype("category")
df["Language"] = df["Language"].astype("category")

print("DataFrame:")
print(df)

print("\nData types:")
print(df.dtypes)

# 3. Display categories in Genre
print("\nCategories in Genre:")
print(df["Genre"].cat.categories)

# 4. Number of movies in each genre
print("\nNumber of movies in each genre:")
print(df["Genre"].value_counts())

# 5. Movies belonging to Action or Drama
print("\nAction or Drama movies:")
print(df[df["Genre"].isin(["Action", "Drama"])])

# 6. Most frequently occurring language
print("\nLanguage frequency:")
print(df["Language"].value_counts())

print("\nMost frequently occurring language:")
print(df["Language"].value_counts().idxmax())

```

```python

import pandas as pd

# 1. Create the DataFrame
data = {
    "Customer_ID": ["C201", "C202", "C203", "C204", "C205",
                    "C206", "C207", "C208", "C209", "C210"],
    "Age": [21, 34, 29, 45, 37, 26, 51, 32, 28, 43],
    "City": ["Hyderabad", "Chennai", "Hyderabad", "Bengaluru", "Chennai",
             "Hyderabad", "Bengaluru", "Hyderabad", "Chennai", "Bengaluru"],
    "Product_Category": ["Electronics", "Clothing", "Electronics", "Furniture",
                         "Clothing", "Grocery", "Electronics", "Furniture",
                         "Grocery", "Clothing"],
    "Purchase_Amount": [18000, 7500, 42000, 68000, 15000,
                        5500, 85000, 48000, 8000, 25000],
    "Rating": [4, 3, 5, 4, 5, 2, 5, 4, 3, 5]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# 2. Customers who spent > 20000 AND have rating 4 or 5
print("\nCustomers with purchase > ₹20,000 and rating 4 or 5:")
print(df[(df["Purchase_Amount"] > 20000) &
         (df["Rating"].isin([4, 5]))])

# 3 & 4. Binning Purchase_Amount
bins = [0, 10000, 40000, float("inf")]
labels = ["Low", "Medium", "High"]

df["Purchase_Level"] = pd.cut(
    df["Purchase_Amount"],
    bins=bins,
    labels=labels,
    right=False
)

print("\nDataFrame after binning:")
print(df)

# 5. Convert City and Product_Category to categorical
df["City"] = df["City"].astype("category")
df["Product_Category"] = df["Product_Category"].astype("category")

print("\nData types:")
print(df.dtypes)

# 6. Display categories in Product_Category
print("\nProduct categories:")
print(df["Product_Category"].cat.categories)

# 7. Number of customers in each Purchase_Level
print("\nNumber of customers in each Purchase Level:")
print(df["Purchase_Level"].value_counts())

# 8. Hyderabad or Bengaluru AND Purchase_Level is High
print("\nCustomers from Hyderabad or Bengaluru with High Purchase Level:")
print(
    df[
        df["City"].isin(["Hyderabad", "Bengaluru"]) &
        (df["Purchase_Level"] == "High")
    ]
)

# 9. Most frequently purchased product category
print("\nProduct category frequency:")
print(df["Product_Category"].value_counts())

print("\nMost frequently purchased product category:")
print(df["Product_Category"].value_counts().idxmax())
```
